# Fehlende Fußgängerüberwege aus Mapillary-Fahrbahnmarkierungen → MapRoulette

Findet Zebrastreifen-Markierungen (`marking--discrete--crosswalk-zebra`), in deren Nähe
OpenStreetMap keinen Überweg (`highway=crossing`) kennt, und macht daraus Aufgaben für die
MapRoulette-Challenge 53697. Läuft von Hand, wenn eine neue Runde fällig ist.

Der Zwilling der Verkehrszeichen-Kampagne
([`../pedestrian_crossing_campaign/maproulette_tasks.ipynb`](../pedestrian_crossing_campaign/maproulette_tasks.ipynb)):
derselbe Abgleich, dieselben Abstände, dasselbe Modul [`../pc_campaign.py`](../pc_campaign.py) —
nur Map-Feature-Punkte statt Verkehrszeichen und ein eigener Aufgabentext.

**Ablauf**

| Schritt | Funktion |
| --- | --- |
| Map-Feature-Punkte von data.vizsim.de spiegeln | `campaign.sync_features` |
| Zebrastreifen einlesen | `campaign.load_features` |
| auf Deutschland beschneiden | `campaign.clip_to_boundary` |
| frische / kurzlebige Markierungen aussortieren | `campaign.filter_stable_signs` |
| Abstand zum nächsten OSM-Überweg | `campaign.distance_to_nearest` |
| Markierungen derselben Stelle zusammenfassen | `campaign.dedupe_nearby` |
| bereits bestehende Aufgaben abziehen | `campaign.drop_near_existing_tasks`, `campaign.drop_known_task_names` |
| **neuestes** Mapillary-Bild je Markierung holen | `campaign.newest_image_ids` |
| GeoJSON für MapRoulette schreiben | `campaign.build_maproulette_geojson` mit `pc.aufgabe_markierung` |
| in die Challenge hochladen (nur mit `CAMPAIGN_UPLOAD=1`) | `campaign.upload_tasks` |

Die Challenge holt ihre Aufgaben aus `maproulette_tasks_missing-ped-crossing_map-feature-point.geojson`
auf GitHub: Notebook laufen lassen, Datei committen und pushen, in MapRoulette „Rebuild tasks“.
Abschnitt 9 lädt stattdessen direkt über die API hoch — sobald MapRoulette Skripten wieder
einen Zugang gibt (siehe [README](README.md)).

## Einstellungen

````python
import json
import os
import sys
from pathlib import Path

import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd

# campaign.py and pc_campaign.py live one level up. ".." works because Jupyter and
# nbconvert run the kernel in the notebook's directory - the same assumption as
# ../../output/ and ../utils/.
sys.path.insert(0, "..")
import campaign
import pc_campaign as pc

# Must match the date in ../0_prepare_osm_network.ipynb, which writes the crossings file.
set_date = "261017"

ordner_punkte = Path("../../output")
pfad_ueberwege = Path(f"../utils/processed_osm_files/processed_crossings_germany_{set_date}.parquet")
pfad_grenze = Path("../utils/OSMB-germany.geojson.gz")
pfad_config = Path("../utils/config_mapillary_privat.json")

# Only markings last seen after this date ...
zuletzt_gesehen_nach = "2025-01-01"
# ... and seen over at least this many months - filters out temporary markings.
mindest_standzeit_monate = 12

challenge_id = 53697

# MapRoulette pulls the tasks from this file on GitHub (remoteGeoJson), so it keeps
# its name. Once uploads by script work again, section 9 replaces commit, push and
# "Rebuild tasks".
ziel_geojson = Path("maproulette_tasks_missing-ped-crossing_map-feature-point.geojson")

# From the environment, so a run on the server needs no edit to this notebook:
#   CAMPAIGN_SYNC=0    read ../../output/ as it is. On the server it is what the
#                      weekly run produced, and mirroring would overwrite its manifest.
#   CAMPAIGN_UPLOAD=1  upload the tasks to the challenge - refused by MapRoulette
#                      until it has tokens for scripts, see the README. Without it,
#                      section 9 only says how many it would upload.
sync = os.environ.get("CAMPAIGN_SYNC", "1") != "0"
hochladen = os.environ.get("CAMPAIGN_UPLOAD") == "1"

# Only the columns needed - the map-feature parquets hold every feature class.
spalten = ["id", "value", "first_seen_at", "last_seen_at", "geometry"]

print("Markierungen:", ", ".join(pc.UEBERWEG_MARKIERUNGEN))
print(f"Suchradius {pc.SUCHRADIUS_M:.0f} m, zusammenfassen bis {pc.ZUSAMMENFASSEN_M:.0f} m")
print(f"spiegeln: {sync}, hochladen: {hochladen}")
````

## 1 · Markierungen holen und einlesen

````python
if sync:
    # Fetches only what is missing locally or newer on data.vizsim.de.
    campaign.sync_features(ordner_punkte, prefix=campaign.PREFIX_MARKIERUNGEN)

manifest = json.loads((ordner_punkte / "ml-mf_metadata.json").read_text(encoding="utf-8"))
print("Datenstand:", manifest["processed_date"])
print("im letzten Lauf unvollständig:", manifest.get("last_run_incomplete") or "nichts")
````

````python
# Expected number of states from the tile cache; None without one.
erwartet = campaign.count_expected_states("../../prep/tile_cache/DE-*_tiles.json")
print("Tile-Cache kennt", erwartet, "Bundesländer")

marks = campaign.load_features(
    ordner_punkte,
    prefix=campaign.PREFIX_MARKIERUNGEN,
    values=pc.UEBERWEG_MARKIERUNGEN,
    columns=spalten,
    expect_files=erwartet,
)
# The parquets are cut along zoom-14 tiles, and border tiles reach abroad - where
# the OSM extract has no crossings, so every marking there would become a task.
marks = campaign.clip_to_boundary(marks, campaign.load_boundary(pfad_grenze))
````

## 2 · Zeitliche Filter

````python
fig, achsen = plt.subplots(1, 2, figsize=(14, 3.5))
marks["last_seen_at"].str[:7].value_counts().sort_index().plot(kind="bar", ax=achsen[0], title="alle Zebrastreifen")

marks = campaign.filter_stable_signs(marks, zuletzt_gesehen_nach, mindest_standzeit_monate)

marks["last_seen_at"].str[:7].value_counts().sort_index().plot(
    kind="bar", ax=achsen[1], title=f"zuletzt gesehen nach {zuletzt_gesehen_nach}, ≥ {mindest_standzeit_monate} Monate"
)
plt.tight_layout()
````

## 3 · Abstand zum nächsten OSM-Überweg

Gemessen wird bis `pc.MESSRADIUS_M` (100 m), damit der Aufgabentext sagen kann, wie weit der
nächste gemappte Überweg entfernt ist. Zur Aufgabe wird eine Markierung nur, wenn innerhalb von
`pc.SUCHRADIUS_M` (50 m) keiner liegt.

````python
ueberwege = gpd.read_parquet(pfad_ueberwege, columns=["geometry"])
print(f"OSM-Überwege (highway=crossing): {len(ueberwege):,}".replace(",", "."))

marks["dist_crossing_m"] = campaign.distance_to_nearest(marks, ueberwege, pc.MESSRADIUS_M)
del ueberwege

print(f"ohne OSM-Überweg in {pc.SUCHRADIUS_M:.0f} m:  {(marks.dist_crossing_m >= pc.SUCHRADIUS_M).sum():>6}")
print(f"ohne OSM-Überweg in {pc.MESSRADIUS_M:.0f} m: {(marks.dist_crossing_m == float('inf')).sum():>6}")
````

## 4 · Kandidaten

Ein Zebrastreifen wird oft mehrfach erkannt, aus beiden Fahrtrichtungen oder auf mehreren
Fahrstreifen. Markierungen, die höchstens `pc.ZUSAMMENFASSEN_M` (100 m) auseinander liegen,
werden zu einer Aufgabe; es bleibt die zuletzt gesehene. Dieselbe Regel wie in der Runde 2025.

````python
kandidaten = marks[marks["dist_crossing_m"] >= pc.SUCHRADIUS_M]
kandidaten = campaign.dedupe_nearby(kandidaten, pc.ZUSAMMENFASSEN_M).reset_index(drop=True)

print(f"Kandidaten: {len(kandidaten)}")
kandidaten.plot(figsize=(6, 8), markersize=4)
````

## 5 · Bereits bestehende Aufgaben abziehen

Alles außer Fixed, Created und Skipped blockiert eine Neuanlage in 50 m. Außerdem fällt jede
Markierung weg, deren ID Name einer erledigten Aufgabe ist: MapRoulette überschriebe die alte
Aufgabe, statt eine neue anzulegen, und sie bliebe erledigt. Ist die Aufgabe dagegen noch
offen, bleibt die Markierung drin — dann aktualisiert der Rebuild sie.

````python
api_key = campaign.load_token(pfad_config, "API_KEY_MAPROULETTE")

# Everything but Fixed / Created / Skipped blocks a new task nearby: what was ticked
# off as "false positive" or "already fixed" must not come back.
bestehende = campaign.load_challenge_tasks(challenge_id, api_key)
print(f"blockierende Aufgaben in Challenge {challenge_id}: {len(bestehende)}")
kandidaten = campaign.drop_near_existing_tasks(kandidaten, bestehende, radius_m=pc.SUCHRADIUS_M)

# A marking whose id names a closed task would only update that task - it would stay
# closed. Open tasks of the same name stay in: the rebuild refreshes them.
kandidaten = campaign.drop_known_task_names(kandidaten, campaign.challenge_task_names(challenge_id, api_key))
print(f"neu anzulegen: {len(kandidaten)}")
````

## 6 · Neuestes Mapillary-Bild

Die Runde 2025 suchte das Bild über Geo-Suchen und den Blickwinkel der Kamera, mit mehreren
Anfragen je Markierung. Die Graph-API liefert aber direkt die Bilder, in denen die Markierung
erkannt wurde (`fields=id,images{id,captured_at}`); das neueste davon zeigt sie sicher. Bis zu
50 Markierungen gehen in eine Anfrage.

````python
bilder = campaign.newest_image_ids(kandidaten["id"], campaign.load_token(pfad_config))

# Join on the string index: Mapillary ids exceed 2**53 and would be rounded as numbers.
kandidaten = kandidaten.set_index(kandidaten["id"].astype(str)).join(bilder).reset_index(drop=True)

kandidaten[["id", "last_seen_at", "dist_crossing_m", "image_id", "image_captured_at"]].head()
````

### Plausibilitätsprüfung

`last_seen_at` aus dem Parquet ist die letzte Aufnahme, auf der die Markierung erkannt wurde.
Stimmt die Bildauswahl, ist das verlinkte Bild ungefähr von diesem Tag.

````python
abstand_tage = (
    kandidaten["image_captured_at"].dt.tz_localize(None) - pd.to_datetime(kandidaten["last_seen_at"])
).dt.days.abs()

print(abstand_tage.describe().to_string())
print(f"\nBild am selben Tag wie last_seen_at: {(abstand_tage <= 1).sum()} von {abstand_tage.notna().sum()}")

if abstand_tage.notna().any():
    abstand_tage.plot(kind="hist", bins=40, figsize=(8, 3), title="Tage zwischen verlinktem Bild und last_seen_at")
````

## 7 · GeoJSON für MapRoulette schreiben

Die Datei behält ihren Namen, weil MapRoulette sie von GitHub holt. Ihr bisheriger Inhalt wird
**vor** dem Überschreiben eingelesen — sonst gäbe es im nächsten Schritt nichts zu vergleichen.

````python
vorheriger_stand = campaign.read_geojson(ziel_geojson)

aufgaben = campaign.build_maproulette_geojson(kandidaten, pc.aufgabe_markierung)
campaign.write_geojson(aufgaben, ziel_geojson)

if aufgaben["features"]:
    print(aufgaben["features"][0]["properties"]["instruction"])
````

## 8 · Vergleich mit dem vorherigen Stand

Stellt die Aufgaben-IDs des bisherigen Datei-Inhalts denen dieses Laufs gegenüber. **anderes
Bild** zählt, wie viele Aufgaben vorher ein anderes Foto verlinkten.

````python
if vorheriger_stand:
    campaign.print_comparison(campaign.compare_task_sets(vorheriger_stand, aufgaben))
else:
    print("kein vorheriger Stand vorhanden - Vergleich übersprungen")
````

## 9 · In die Challenge hochladen

Direkt über die MapRoulette-API (`addFileTasks`), ohne Commit und ohne „Rebuild tasks“. Die
Aufgaben, die schon in der Challenge stehen, bleiben, wie sie sind. Danach prüft
`upload_tasks`, ob jede Aufgabe angekommen ist — MapRoulette meldet Erfolg auch dann, wenn
der Import intern scheitert.

Läuft nur mit `CAMPAIGN_UPLOAD=1`, und **noch gar nicht**: Seit maproulette-backend v4.11.0
(Oktober 2026) lehnt MapRoulette API-Keys ab; der angekündigte Ersatz, Personal Access Tokens,
fehlt noch. Bis dahin: Datei committen, pushen, „Rebuild tasks“.

````python
if hochladen:
    nachher = campaign.upload_tasks(
        challenge_id, aufgaben, api_key, data_origin_date=manifest["processed_date"][:10]
    )
    print(pd.Series(nachher).value_counts().to_string())
else:
    print(f"nicht hochgeladen ({len(aufgaben['features'])} Aufgaben) - dafür CAMPAIGN_UPLOAD=1 setzen")
````

## Challenge-Beschreibung (für MapRoulette)

## 🚶 Fehlende Fußgängerüberwege anhand von Mapillary Map Feature Points ergänzen (Deutschland)

Diese Challenge basiert auf automatisch erkannten **Map Feature Points** aus Mapillary-Bildern in Deutschland.

### 📌 Kriterien für jede Aufgabe

Nur Aufgaben, die **alle** folgenden Bedingungen erfüllen, wurden berücksichtigt:

- Das Map Feature wurde **in Mapillary erkannt**.
- Es handelt sich um ein **Lane Marking - Crosswalk**
  *(„Zebrastreifen“).*
- Die Markierung wurde **mindestens 12 Monate lang regelmäßig erkannt**.
- Die neueste Aufnahme stammt **aus dem Jahr 2025 oder später**.
- In OpenStreetMap befindet sich **kein Fußgängerüberweg** (`highway=crossing`) **innerhalb von 50 m** des erkannten Standortes.
- Mehrere Markierungen an derselben Stelle ergeben **eine** Aufgabe.

---

### 🔍 Was du tun solltest

1. Öffne den Ort in **Mapillary** (über den Link in der Aufgabe).
2. Prüfe, ob dort tatsächlich ein **Fußgängerüberweg** existiert.
3. Falls ja, ergänze in OSM die passenden Tags, z. B.:

```
highway=crossing
crossing=uncontrolled
crossing:markings=zebra
crossing:island=no
```
*(ggf. auch `kerb=lowered`, `tactile_paving=yes`, etc.)*

4. Wenn bereits ein Fußgängerüberweg korrekt gemappt ist, kannst du die Aufgabe **als erledigt markieren**.

---

### 💡 Hinweise

- Mapillary-Bilder können veraltet oder ungenau positioniert sein – prüfe daher **immer den aktuellen Stand**.
- Fußgängerüberwege werden in OSM als **Node auf der Straßenlinie** (`highway=*`) eingetragen, nicht als eigener Weg.
- Falls kein Überweg existiert oder unklar ist, kannst du die Aufgabe **überspringen**.

---

🗺️ **Vielen Dank**, dass du hilfst, die Fußgängerinfrastruktur in OpenStreetMap zu verbessern!
